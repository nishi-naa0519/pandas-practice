```python
import pandas as pd
from src.grader import check, hint

df = pd.read_csv("data/app_users.csv")
```

## 1. 模範解答

```python
def solve(df: pd.DataFrame):
    return df["os"].str.lower().value_counts()
```

```python
solve(df)
```

```python
check("P032", solve, record=False)  # 解答例の確認用(履歴には記録しない)
```

## 2. コードの説明

`.str.lower()` で列のすべての値を小文字にし、続けて `value_counts()` で数えます。

## 3. 今回覚えるポイント

- カテゴリ列は `value_counts()` で**表記ゆれがないか**を必ず見る
- 小文字にそろえる: `df["col"].str.lower()`
- 前後の空白(`"Tokyo "` など)もよくある表記ゆれ。見つけたら AI に整形を頼んでもよい

## 4. よくある間違い

- `df["os"].lower()` → `.str` が必要
- `value_counts()` してから小文字にしても、件数はまとまらない
