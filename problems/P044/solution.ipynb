```python
import pandas as pd
from src.grader import check, hint

df = pd.read_csv("data/churn_train.csv")
```

## 1. 模範解答

```python
def solve(df: pd.DataFrame):
    missing_rate = df.isna().mean()
    n_unique = df.nunique()
    is_candidate = (missing_rate > 0.5) | (n_unique == 1)
    return list(df.columns[is_candidate])
```

```python
solve(df)
```

```python
check("P044", solve, record=False)  # 解答例の確認用(履歴には記録しない)
```

## 2. コードの説明

1. `df.isna().mean()` で列ごとの欠損率、`df.nunique()` で列ごとのユニーク数
2. どちらも index が列名の Series なので、`|` で「どちらかを満たす」を作れる
3. `df.columns[is_candidate]` で True の列名だけを取り出す

すべて欠損している列はユニーク数が 0 になりますが、欠損率の条件で拾えます。

## 3. 今回覚えるポイント

- 欠損率 `df.isna().mean()` とユニーク数 `df.nunique()` は、列の要約として最もよく使う2つ
- 同じ index の Series どうしは `&` `|` で組み合わせられる
- このデータでは `coupon_code`(約87%欠損)と `country`(全部 "JP")が該当する
- あくまで「候補」。欠損していること自体が情報になる場合もある(クーポンを持っているか、など)

## 4. よくある間違い

- `>= 0.5` にしてしまう → 「半分を超えて」なので `> 0.5`
- `n_unique <= 1` にすると、全部欠損の列も拾えるが条件の意味が変わる。今回はどちらでも結果は同じ
