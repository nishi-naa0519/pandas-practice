**解答と解説**

```python
import pandas as pd
from src.grader import check, hint

orders = pd.read_csv("data/orders.csv")
customers = pd.read_csv("data/customers.csv")
```

## 1. 模範解答

```python
def solve(orders: pd.DataFrame, customers: pd.DataFrame):
    merged = pd.merge(orders, customers, on="customer_id", how="left")
    return merged["plan"].isna().sum()
```

```python
solve(orders, customers)
```

```python
check("P036", solve, record=False)  # 解答例の確認用(履歴には記録しない)
```

## 2. コードの説明

`how="left"` にすると、`orders` の行はすべて残り、対応する顧客がいない行では `customers` 側の列が欠損になります。
customers 側で必ず値が入っている列(`plan`)の欠損を数えれば、対応のない注文の件数が分かります。

## 3. 今回覚えるポイント

- left join: `pd.merge(left, right, on="key", how="left")`
- 対応のない行の数 = 右側の「本来は欠損しない列」の欠損数
- `age` のように元から欠損がある列で判定すると、件数が多めに出てしまう

## 4. よくある間違い

- `how="left"` を忘れる → inner join になり、対応のない行は消えて 0 件になる
- `merged.isna().sum().sum()` → 全列の欠損の合計で、件数にならない
