**解答と解説**

```python
import pandas as pd
from src.grader import check, hint

df = pd.read_csv("data/web_logs.csv")
```

## 1. 模範解答

```python
def solve(df: pd.DataFrame):
    return df.loc[df["converted"] == 1, ["session_id", "page", "duration_sec"]]
```

```python
solve(df)
```

```python
check("P012", solve, record=False)  # 解答例の確認用(履歴には記録しない)
```

## 2. コードの説明

`df.loc[行, 列]` の形で、行には条件(True/False の Series)、列には列名のリストを渡します。

## 3. 今回覚えるポイント

- `df.loc[条件, ["列1", "列2"]]` で「条件の行の、この列だけ」
- 列を全部残すなら `df[条件]` で十分

## 4. よくある間違い

- `df.loc[条件, "session_id", "page"]` → 列はリストにまとめる
- `df.iloc[条件, ...]` → iloc は位置指定用。条件や列名には loc を使う
