```python
import pandas as pd
from src.grader import check, hint

df = pd.read_csv("data/customers.csv")
```

## 1. 模範解答

```python
def solve(df: pd.DataFrame):
    return df[df["age"].isna() | df["prefecture"].isna()]
```

```python
solve(df)
```

```python
check("P016", solve, record=False)  # 解答例の確認用(履歴には記録しない)
```

## 2. コードの説明

`df["age"].isna()` と `df["prefecture"].isna()` はどちらも True/False の Series です。
`|` で「どちらかが True」にして、boolean indexing で行を絞ります。
(`.isna()` はメソッド呼び出しなので、この形なら `()` で囲まなくても動きます。比較演算子を使うときは囲みます)

## 3. 今回覚えるポイント

- 欠損している行を見る: `df[df["age"].isna()]`
- 欠損していない行は `df[~df["age"].isna()]`
- 件数だけでなく**欠損している行の中身**を見ると、「特定の期間・グループだけ欠損」などに気づける

## 4. よくある間違い

- `df[df.isna()]` → 行の抽出にならない(欠損以外が NaN になった表が返る)
- `df["age"] == None` や `df["age"] == np.nan` → 欠損の判定には使えない。必ず `isna()`
