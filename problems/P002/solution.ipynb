## 2. コードの説明

`df.shape` は `(行数, 列数)` のタプルです。メソッドではなく**属性**なので括弧を付けません。

## 3. 今回覚えるポイント

- `df.shape` → `(行数, 列数)`
- 行数だけなら `df.shape[0]` か `len(df)`
- train / test を受け取ったら、まず両方の `shape` を見る習慣をつける

## 4. よくある間違い

- `df.shape()` → `TypeError: 'tuple' object is not callable`(属性なので括弧不要)
- `len(df)` は行数だけ。列数は分かりません
