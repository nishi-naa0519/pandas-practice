```python
import pandas as pd
from src.grader import check, hint

df = pd.read_csv("data/customers.csv")
```

## 1. 模範解答

```python
def solve(df: pd.DataFrame):
    return df[["customer_id", "age", "plan"]]
```

```python
solve(df)
```

```python
check("P007", solve, record=False)  # 解答例の確認用(履歴には記録しない)
```

## 2. コードの説明

`df[["customer_id", "age", "plan"]]` のように**列名のリスト**を渡すと、その列だけの DataFrame になります。
列の順番はリストの順番になります。

## 3. 今回覚えるポイント

- 1列: `df["age"]` → Series
- 複数列: `df[["a", "b"]]` → DataFrame(外側の `[]` が選択、内側の `[]` がリスト)

## 4. よくある間違い

- `df["customer_id", "age"]` → `KeyError`(リストにしていない)
- `df[["customer_id"], ["age"]]` → リストは1つにまとめる
