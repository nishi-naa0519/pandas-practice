**解答と解説**

```python
import pandas as pd
from src.grader import check, hint

df = pd.read_csv("data/web_logs.csv")
```

## 1. 模範解答

```python
def solve(df: pd.DataFrame):
    ts = pd.to_datetime(df["timestamp"])
    return (ts.max() - ts.min()).days
```

```python
solve(df)
```

```python
check("P040", solve, record=False)  # 解答例の確認用(履歴には記録しない)
```

## 2. コードの説明

1. `pd.to_datetime()` で日時型にする
2. `.max()` と `.min()` で最後と最初の日時を取る(並び順に関係なく正しく取れる)
3. 引き算すると Timedelta(期間)になり、`.days` で日数(切り捨て)が取れる

## 3. 今回覚えるポイント

- 期間の確認: `ts.min(), ts.max()` を見るだけでも十分。新しいデータを受け取ったら必ず見る
- 日時の差は Timedelta。`.days` で日数
- 文字列のままでも min / max は動くが、**日付の形式がそろっていないと誤る**。変換してから比べる

## 4. よくある間違い

- `df["timestamp"].iloc[-1] - df["timestamp"].iloc[0]` → 並び順に依存する。ソートされていないと誤る
- `.days()` → 括弧は付けない
