**解答と解説**

```python
import pandas as pd
from src.grader import check, hint

df = pd.read_csv("data/churn_train.csv")
```

## 1. 模範解答

```python
def solve(df: pd.DataFrame):
    df = df.drop_duplicates()
    df = df.dropna(subset=["churn"])
    return df
```

```python
solve(df)
```

```python
check("P034", solve, record=False)  # 解答例の確認用(履歴には記録しない)
```

## 2. コードの説明

- `drop_duplicates()` は完全に重複した行の2回目以降を削除します
- `dropna(subset=["churn"])` は churn が欠損している行だけを削除します

## 3. 今回覚えるポイント

- 重複行の削除: `df.drop_duplicates()`
- 目的変数が欠損した行の削除: `df.dropna(subset=["target"])`
- `subset` を付けない `dropna()` は**どこか1列でも欠損があれば削除**する。特徴量の欠損が多いデータではほとんどの行が消えることも

## 4. よくある間違い

- `df.dropna()` → 他の列の欠損まで削除してしまう(このデータでは `coupon_code` がほとんど欠損しているため、604 行が 51 行まで減る)
- `df.drop_duplicates()` だけで代入しない → df は変わらない
