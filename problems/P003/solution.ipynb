## 2. コードの説明

`df.columns` は列名の一覧(Index)です。リストが欲しいときは `list(df.columns)` にします。

## 3. 今回覚えるポイント

- 列名の確認は `df.columns`
- 列名が多くて表示が省略されるときは `list(df.columns)` にすると全部見える
- 列名に空白や大文字小文字の違いがないかもここで確認する

## 4. よくある間違い

- `df.columns()` → エラー(属性なので括弧不要)
- `df.column` → 綴り間違い(複数形の `columns`)
