```python
import pandas as pd
from src.grader import check, hint

df = pd.read_csv("data/customers.csv")
```

## 1. 模範解答

```python
def solve(df: pd.DataFrame):
    df["age"] = df["age"].fillna(df["age"].median())
    return df
```

```python
solve(df)
```

```python
check("P029", solve, record=False)  # 解答例の確認用(履歴には記録しない)
```

## 2. コードの説明

1. `df["age"].median()` で中央値を計算(欠損は除いて計算される)
2. `df["age"].fillna(中央値)` で欠損を埋めた Series を作る
3. `df["age"] = ...` で元の列に代入し直す

## 3. 今回覚えるポイント

- 1列の欠損を埋める: `df["col"] = df["col"].fillna(値)`
- `fillna` は新しい Series を返すだけ。**代入しないと df は変わらない**
- 中央値は極端な値に強いので、年齢や金額の穴埋めによく使う

## 4. よくある間違い

- `df.fillna(df["age"].median())` → 全列の欠損がその値で埋まる(prefecture まで 30 などになる)
- `df["age"].fillna(..., inplace=True)` → pandas 3 では元の df に反映されない
