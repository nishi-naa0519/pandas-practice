## 3. このデータで観察すべきポイント

1. **250行 × 8列**
2. 欠損があるのは **`avg_session_min` だけ**(19件)。`info()` の Non-Null Count が 231 になっている
3. `str` の列: `user_id`, `os`, `app_version`, `country`, `signup_date`
   - `signup_date` は日付だが文字列のまま(期間を計算するなら `pd.to_datetime` が必要)
   - `app_version` は `"3.2.1"` のような形式なので文字列で正しい
4. 値の範囲
   - `sessions_7d` の **max が 999**。平均 13.7 に対して中央値(50%)は 6、std は 80 と異常に大きい → 極端な値が混ざっている
   - `avg_session_min` の **min が -3.0**。利用時間が負になることはない
5. `user_id` 順・`signup_date` 順(2024-04-01 〜 2025-03-31)に並んでいる。`head()` だけ見ると古いユーザーしか見えない

※ `os` 列の表記ゆれ(`iOS` / `ios` / `IOS`)はここでは見つからなくても大丈夫です(P032 で扱います)。

## 4. 今回覚えるポイント

- 初見のデータでは **shape → head → info → describe** を流れで打つ
- `info()` は型と欠損を一度に見られる(戻り値はなく、表示するだけ)
- `describe()` では **mean と 50% の差、min と max** を必ず見る
- `head()` だけでは並び順の偏りに気づけない。`tail()` や `sample()` も見る
