**解答と解説**

```python
import pandas as pd
from src.grader import check, hint

df = pd.read_csv("data/store_sales.csv")
```

## 1. 模範解答

```python
def solve(df: pd.DataFrame):
    return df.groupby(["region", "product"])["units"].sum().reset_index()
```

```python
solve(df)
```

```python
check("P025", solve, record=False)  # 解答例の確認用(履歴には記録しない)
```

## 2. コードの説明

- `groupby(["region", "product"])` のようにリストで渡すと、組み合わせごとに集計します
- そのままだと index が「region と product の2段」(MultiIndex)になって扱いにくいので、
  `.reset_index()` で普通の列に戻します

## 3. 今回覚えるポイント

- 複数キー: `df.groupby(["a", "b"])["val"].sum().reset_index()`
- MultiIndex は読めればよい。自分で扱うときは `reset_index()` で普通の表にしてしまう

## 4. よくある間違い

- `groupby("region", "product")` → リストにしていない
- `reset_index()` を忘れる → index が2段の Series のまま
