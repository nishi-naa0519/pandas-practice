```python
import pandas as pd
from src.grader import check, hint

df = pd.read_csv("data/orders.csv")
```

## 1. 模範解答

```python
def solve(df: pd.DataFrame):
    return df["payment_method"].value_counts()
```

```python
solve(df)
```

```python
check("P019", solve, record=False)  # 解答例の確認用(履歴には記録しない)
```

## 2. コードの説明

`value_counts()` は値ごとの件数を数え、**多い順**に並べて返します。欠損は数えません。

## 3. 今回覚えるポイント

- カテゴリの分布: `df["col"].value_counts()`
- 割合で見たいときは `value_counts(normalize=True)`(次の問題)
- 欠損も1つの値として数えたいときは `value_counts(dropna=False)`

## 4. よくある間違い

- `df.value_counts()` → 全列の組み合わせの件数になる。列を選んでから使う
- `df.groupby("payment_method").count()` → 列ごとの非欠損数が並ぶ表になり、見づらい
