````python
import pandas as pd
from src.grader import check, hint

df = pd.read_csv("data/orders.csv")
````

## 1. 模範解答

````python
def solve(df: pd.DataFrame):
    return df.groupby("category")["price"].mean()
````

````python
solve(df)
````

````python
check("P023", solve, record=False)  # 解答例の確認用(履歴には記録しない)
````

## 2. コードの説明

`groupby` は「**どの列でグループ分けし** → **どの列を** → **どう集計するか**」の順に書きます。

```python
df.groupby("category")["price"].mean()
#        ↑グループの列   ↑集計する列  ↑集計方法
```

欠損は除いて平均されます。

## 3. 今回覚えるポイント

- 基本形: `df.groupby("cat")["val"].mean()`
- `.mean()` を `.sum()` `.median()` `.max()` `.count()` に変えるだけで他の集計になる
- 結果の index はグループの値(アルファベット順に並ぶ)

## 4. よくある間違い

- `df.groupby("category").mean()` → 全部の数値列の平均になり、文字列列があるとエラーになる。集計する列を選ぶ
- `df.groupby(["price"])["category"]...` → グループの列と集計する列が逆
