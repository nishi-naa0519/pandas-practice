```python
import pandas as pd
from src.grader import check, hint

df = pd.read_csv("data/orders.csv")
```

## 1. 模範解答

```python
def solve(df: pd.DataFrame):
    return df[df["price"] >= 3000]
```

```python
solve(df)
```

```python
check("P008", solve, record=False)  # 解答例の確認用(履歴には記録しない)
```

## 2. コードの説明

1. `df["price"] >= 3000` で、行ごとに True/False の Series を作ります
2. それを `df[...]` に渡すと、True の行だけが残ります(boolean indexing)

## 3. 今回覚えるポイント

- 条件抽出の基本形は `df[df["col"] 比較 値]`
- 比較演算子: `==`, `!=`, `>`, `>=`, `<`, `<=`
- 抽出後も元の index(行ラベル)が残る。番号が飛んでいても正常
- 件数だけ知りたいときは `(df["price"] >= 3000).sum()`

## 4. よくある間違い

- `df["price" >= 3000]` → 括弧の位置が違う
- `>`(より大きい)と `>=`(以上)の取り違え
