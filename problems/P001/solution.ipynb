```python
import pandas as pd
from src.grader import check, hint

path = "data/orders.csv"
```

## 1. 模範解答

```python
def solve(path: str):
    df = pd.read_csv(path)
    return df.head(3)
```

```python
solve(path)
```

```python
check("P001", solve, record=False)  # 解答例の確認用(履歴には記録しない)
```

## 2. コードの説明

- `pd.read_csv(path)` でCSVを DataFrame として読み込みます。
- `df.head(3)` で先頭3行を取り出します。引数を省略すると5行です。

## 3. 今回覚えるポイント

- 読み込みは `pd.read_csv("ファイルパス")`
- 中身の確認は `df.head()`(末尾は `df.tail()`)
- Notebook ではセルの最後の式が自動で表示されるので、普段は `df.head()` と書くだけでOK

## 4. よくある間違い

- `df.head` と括弧を忘れる → メソッドそのものが表示されるだけで、データは出ません
- `df.head[3]` と角括弧にする → エラー。メソッドの引数は `()` で渡します
- 行数が3行未満のデータでも `head(3)` はエラーにならず、ある分だけ返します
