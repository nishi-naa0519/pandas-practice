```python
import pandas as pd
from src.grader import check, hint

df = pd.read_csv("data/customers.csv")
```

## 1. 模範解答

```python
def solve(df: pd.DataFrame):
    return df["customer_id"].nunique() == len(df)
```

```python
solve(df)
```

```python
check("P018", solve, record=False)  # 解答例の確認用(履歴には記録しない)
```

## 2. コードの説明

`df["customer_id"].nunique()` は ID の種類数です。行数 `len(df)` と同じなら、すべての行で ID が異なります。
このデータでは False になります(重複行が混ざっているため)。

## 3. 今回覚えるポイント

- ID の一意性: `df["id"].nunique() == len(df)`
- 重複している ID の数を知りたいときは `df["id"].duplicated().sum()`
- 「ID が一意でない」と、結合で行が増える・集計が二重になるなどの問題が起きる

## 4. よくある間違い

- `df.nunique()` → 全列のユニーク数になる。1列を選んでから使う
- `df["customer_id"].unique() == len(df)` → unique() は値の配列。数は nunique()
