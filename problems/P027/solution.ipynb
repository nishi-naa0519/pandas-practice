```python
import pandas as pd

df = pd.read_csv("data/web_logs.csv")
```

## 1. 推奨確認項目

1. グループの種類と件数(`value_counts`)
2. グループごとの率と件数(`groupby` + `agg(["mean", "count"])`)
3. 人間以外のアクセスなど、集計対象にすべきでない行がないか(滞在時間などの他の列も見る)
4. 除外後にもう一度集計

## 2. 模範的なコード

```python
df["device"].value_counts()
```

```python
df.groupby("device")["converted"].agg(["mean", "count"])
```

```python
df.groupby("device")["duration_sec"].mean()
```

```python
human = df[~df["device"].isin(["bot", "crawler"])]
human.groupby("device")["converted"].agg(["mean", "count"])
```

```python
# 全体のコンバージョン率(除外前と後)
df["converted"].mean(), human["converted"].mean()
```

## 3. このデータで観察すべきポイント

1. device は mobile(224件)・pc(173)・tablet(35)・**bot(48)・crawler(20)** の5種類
2. コンバージョン率は pc 約9.2%、mobile 約8.0%、tablet 約5.7%、bot と crawler は 0%
3. **bot / crawler は人間のアクセスではない**。滞在時間の平均が約1秒しかない。
   含めたままだと全体のコンバージョン率が 7.2% → 除外すると 8.3% と、実態より低く見える
4. **tablet は35件しかない**。購入は2件だけなので、1件増えるだけで率が大きく変わる。
   「tablet は購入率が低い」と結論するには件数が足りない
5. pc と mobile の差(約1ポイント)も、この件数では偶然の範囲かもしれない。言い切らないこと

## 4. 今回覚えるポイント

- グループ比較は **率と件数をセットで**: `agg(["mean", "count"])`
- 集計の前に「そもそも集計対象にすべきでない行」がないか考える
- 除外は `~df["col"].isin([...])` で書ける
