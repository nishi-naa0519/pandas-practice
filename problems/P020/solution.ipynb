```python
import pandas as pd
from src.grader import check, hint

df = pd.read_csv("data/churn_train.csv")
```

## 1. 模範解答

```python
def solve(df: pd.DataFrame):
    return df["churn"].value_counts(normalize=True)
```

```python
solve(df)
```

```python
check("P020", solve, record=False)  # 解答例の確認用(履歴には記録しない)
```

## 2. コードの説明

`normalize=True` を付けると、件数ではなく**割合**(合計 1)になります。
このデータでは churn=1 が約2割で、クラスが偏っています。

## 3. 今回覚えるポイント

- クラス比率: `df["target"].value_counts(normalize=True)`
- 偏りが大きいと、accuracy が高くても役に立たないモデルになりうる(全部 0 と予測しても約8割正解)
- 実データの `churn` には欠損が少しある。`value_counts(dropna=False)` で欠損も含めて確認しておくとよい

## 4. よくある間違い

- `df["churn"].mean()` → 0/1 なら churn=1 の割合は出るが、クラスが3つ以上だと使えない
- `normalize` の綴り間違い(`normalized` など)
