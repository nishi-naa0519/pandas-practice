```python
import pandas as pd
from src.grader import check, hint

df = pd.read_csv("data/app_users.csv")
```

## 1. 模範解答

```python
def solve(df: pd.DataFrame):
    paid = df[df["is_paid"] == 1]
    return paid["sessions_7d"].median(), paid["sessions_7d"].max()
```

```python
solve(df)
```

```python
check("P022", solve, record=False)  # 解答例の確認用(履歴には記録しない)
```

## 2. コードの説明

1. `df[df["is_paid"] == 1]` で有料ユーザーに絞り、変数 `paid` に入れる
2. `paid["sessions_7d"]` に `.median()` と `.max()` を使う
3. `return a, b` でタプルとして返す

実データでは中央値 10、最大値 19 です。絞り込みを忘れて全ユーザーで計算すると、最大値は 999
(無料ユーザーに混ざっている異常値)になります。

## 3. 今回覚えるポイント

- 基本統計: `.mean()` `.median()` `.std()` `.min()` `.max()` `.sum()` `.count()`
- 絞り込みが入るときは、一度変数に入れると読みやすい
- **平均より中央値**のほうが極端な値に引きずられにくい。両方見て差が大きければ外れ値を疑う

## 4. よくある間違い

- `df["sessions_7d"].median()` → 絞り込みを忘れて全ユーザーの値になる
- `df.median()` → 全列の中央値になり、文字列列でエラーになることもある
