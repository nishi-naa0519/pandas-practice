```python
import pandas as pd
from src.grader import check, hint

df = pd.read_csv("data/orders.csv")
```

## 1. 模範解答

```python
def solve(df: pd.DataFrame):
    return df[(df["category"] == "book") & (df["quantity"] >= 2)]
```

```python
solve(df)
```

```python
check("P009", solve, record=False)  # 解答例の確認用(履歴には記録しない)
```

## 2. コードの説明

- それぞれの条件で True/False の Series を作り、`&` で「両方 True」の行を残します
- `&` は比較演算子より先に計算されるため、**各条件を `()` で囲む必要があります**

## 3. 今回覚えるポイント

- かつ: `&` / または: `|` / 否定: `~`
- 条件は必ず `()` で囲む: `df[(条件1) & (条件2)]`
- 長くなるときは `is_book = df["category"] == "book"` のように変数に分けてもよい

## 4. よくある間違い

- `and` を使う → `ValueError: The truth value of a Series is ambiguous`
- 括弧を忘れる: `df["category"] == "book" & df["quantity"] >= 2` → エラーまたは意図しない結果
- 文字列の比較で `"Book"` のように大文字小文字を間違える
