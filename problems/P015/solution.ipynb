```python
import pandas as pd
from src.grader import check, hint

df = pd.read_csv("data/churn_train.csv")
```

## 1. 模範解答

```python
def solve(df: pd.DataFrame):
    return df.isna().mean().sort_values(ascending=False)
```

```python
solve(df)
```

```python
check("P015", solve, record=False)  # 解答例の確認用(履歴には記録しない)
```

## 2. コードの説明

1. `df.isna()` で欠損が True の表を作る
2. `.mean()` で列ごとの True の割合 = 欠損率
3. `.sort_values(ascending=False)` で大きい順に並べる

## 3. 今回覚えるポイント

- 欠損率: `df.isna().mean()`(`.sum()` を行数で割る必要はない)
- 降順は `ascending=False`
- 欠損率ランキングは、特徴量テーブルを受け取ったら最初に見るものの1つ

## 4. よくある間違い

- `.sum()` のままにする → 欠損数になる
- `sort_values()` だけ → 昇順(小さい順)になる
- `df.isna().mean(axis=1)` → 行ごとの欠損率になってしまう
