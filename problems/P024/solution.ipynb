**解答と解説**

```python
import pandas as pd
from src.grader import check, hint

df = pd.read_csv("data/app_users.csv")
```

## 1. 模範解答

```python
def solve(df: pd.DataFrame):
    return df.groupby("country")["sessions_7d"].mean().sort_values(ascending=False)
```

```python
solve(df)
```

```python
check("P024", solve, record=False)  # 解答例の確認用(履歴には記録しない)
```

## 2. コードの説明

`groupby(...)[...].mean()` の結果は「国 → 平均」の Series です。Series なので `.sort_values()` で並べられます。

## 3. 今回覚えるポイント

- グループ集計 → 並び替え: `df.groupby("cat")["val"].mean().sort_values(ascending=False)`
- 実データでは極端な値(999 など)が平均を大きく動かすことがある。気になったら `.median()` でも見る

## 4. よくある間違い

- `sort_values("sessions_7d")` → Series の sort_values に列名は不要
- 並び替えを groupby の前にしても、集計結果の順番は変わらない
