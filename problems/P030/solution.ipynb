**解答と解説**

```python
import pandas as pd
from src.grader import check, hint

df = pd.read_csv("data/ad_logs.csv")
```

## 1. 模範解答

```python
def solve(df: pd.DataFrame):
    df["ad_id"] = df["ad_id"].astype(str)
    return df
```

```python
solve(df)
```

```python
check("P030", solve, record=False)  # 解答例の確認用(履歴には記録しない)
```

## 2. コードの説明

`.astype(str)` で文字列型に変換し、元の列に代入し直します。

## 3. 今回覚えるポイント

- 型変換: `df["col"] = df["col"].astype(str)`(`int` や `float` も同様)
- ID を数値のままにしておくと、`describe()` に出てきたり、モデルに数値特徴量として入ってしまったりする
- `"1,200"` や `"unknown"` が混ざった列は `astype(int)` ではエラーになる。そういう変換は AI に書かせてよい

## 4. よくある間違い

- `df["ad_id"].astype(str)` だけで代入しない → df は変わらない
- `astype("string")` でも動くが、このカリキュラムでは `astype(str)` に統一する
