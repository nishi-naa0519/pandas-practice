```python
import pandas as pd
from src.grader import check, hint

df = pd.read_csv("data/churn_train.csv")
```

## 1. 模範解答

```python
def solve(df: pd.DataFrame):
    return df.groupby("plan")["churn"].agg(["mean", "count"])
```

```python
solve(df)
```

```python
check("P026", solve, record=False)  # 解答例の確認用(履歴には記録しない)
```

## 2. コードの説明

`.agg(["mean", "count"])` で、1つの列に複数の集計をまとめてかけられます。結果の列名は集計名になります。

0/1 の目的変数の平均は「1 の割合」なので、`mean` がそのまま解約率です。
`count` は欠損を除いた件数です。

## 3. 今回覚えるポイント

- 平均と件数を同時に: `df.groupby("cat")["target"].agg(["mean", "count"])`
- **件数の少ないグループの平均は信用しない**(数件しかないと偶然で大きく変わる)
- `agg` で書くのはこの形まで。列名を自分で付ける書き方(named aggregation)は読めればよい

## 4. よくある間違い

- `.agg("mean", "count")` → リストにする
- `.count()` と `.size()` の違い: count は欠損を除く、size は欠損も含めた行数
