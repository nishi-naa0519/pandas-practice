**解答と解説**

```python
import pandas as pd
from src.grader import check, hint

df = pd.read_csv("data/customers.csv")
```

## 1. 模範解答

```python
def solve(df: pd.DataFrame):
    return df[df["prefecture"].isin(["Tokyo", "Osaka"])]
```

```python
solve(df)
```

```python
check("P011", solve, record=False)  # 解答例の確認用(履歴には記録しない)
```

## 2. コードの説明

`df["prefecture"].isin(["Tokyo", "Osaka"])` は、値がリストのどれかに一致すれば True になります。
欠損(NaN)の行は False になります。

## 3. 今回覚えるポイント

- 複数の値との一致は `.isin([...])`。`==` を `|` で並べるより短く、値が増えても楽
- 否定(リストのどれでもない)は `~df["col"].isin([...])`

## 4. よくある間違い

- `.isin("Tokyo", "Osaka")` → 値はリストで1つの引数として渡す
- `df["prefecture"] == ["Tokyo", "Osaka"]` → エラー
