# P033: 特定の文字列を含む行を集計する

Level 2 ・ コーディング ・ 目安 3 分

**解答と解説**

```python
import pandas as pd
from src.grader import check, hint

df = pd.read_csv("data/web_logs.csv")
```

## 1. 模範解答

```python
def solve(df: pd.DataFrame):
    product = df[df["page"].str.contains("/product/")]
    return product["device"].value_counts()
```

```python
solve(df)
```

```python
check("P033", solve, record=False)  # 解答例の確認用(履歴には記録しない)
```

## 2. コードの説明

1. `df["page"].str.contains("/product/")` で、部分一致する行が True の Series を作る
2. boolean indexing で行を絞る
3. 絞り込んだ DataFrame の `device` 列を `value_counts()`

## 3. 今回覚えるポイント

- 部分一致: `df[df["col"].str.contains("文字列")]`
- 「絞る → 列を選ぶ → 集計」は EDA で最もよく使う組み合わせ

## 4. よくある間違い

- 列に欠損があると `str.contains` の結果に NaN が混ざり、絞り込みでエラーになる(そのときは `na=False` を付ける)
- `df["page"] == "/product/"` → 完全一致になり、`/product/123` は含まれない
