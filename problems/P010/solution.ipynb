**解答と解説**

```python
import pandas as pd
from src.grader import check, hint

df = pd.read_csv("data/customers.csv")
```

## 1. 模範解答

```python
def solve(df: pd.DataFrame):
    return df[(df["plan"] == "premium") | (df["n_orders"] >= 10)]
```

```python
solve(df)
```

```python
check("P010", solve, record=False)  # 解答例の確認用(履歴には記録しない)
```

## 2. コードの説明

`|` は「どちらか一方でも True なら True」です。両方の条件を満たす行も1回だけ出てきます。

## 3. 今回覚えるポイント

- または: `df[(条件1) | (条件2)]`
- 両方満たす行が2回出ることはない(行ごとに True/False を判定しているだけ)

## 4. よくある間違い

- `or` を使う → `ValueError`
- 括弧を忘れる → `|` が先に計算されてエラー
