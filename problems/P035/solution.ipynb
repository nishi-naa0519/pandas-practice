**解答と解説**

```python
import pandas as pd
from src.grader import check, hint

orders = pd.read_csv("data/orders.csv")
customers = pd.read_csv("data/customers.csv")
```

## 1. 模範解答

```python
def solve(orders: pd.DataFrame, customers: pd.DataFrame):
    return pd.merge(orders, customers, on="customer_id")
```

```python
solve(orders, customers)
```

```python
check("P035", solve, record=False)  # 解答例の確認用(履歴には記録しない)
```

## 2. コードの説明

`pd.merge(左, 右, on="キー")` で、キーの値が一致する行どうしをつなげます。
`how` を省略すると inner join になり、**両方に存在するキーの行だけ**が残ります。

## 3. 今回覚えるポイント

- 結合の基本形: `pd.merge(left, right, on="key")`
- inner(デフォルト): 両方にあるキーだけ / left: 左の行はすべて残す(次の問題)
- 結合したら**行数を確認**する(`len(orders)` と比べる)

## 4. よくある間違い

- `pd.merge(customers, orders, ...)` → 結果の列の順番が変わる
- キー名が違う表どうし(`id` と `customer_id` など)は `on` では結合できない(`left_on` / `right_on` は Read 対象)
