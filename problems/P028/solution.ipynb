```python
import pandas as pd
from src.grader import check, hint

df = pd.read_csv("data/orders.csv")
```

## 1. 模範解答

```python
def solve(df: pd.DataFrame):
    df["amount"] = df["price"] * df["quantity"]
    return df
```

```python
solve(df)
```

```python
check("P028", solve, record=False)  # 解答例の確認用(履歴には記録しない)
```

## 2. コードの説明

`df["price"] * df["quantity"]` は、同じ行どうしをまとめて掛け算します(ベクトル演算)。
その結果を `df["amount"] = ...` で新しい列として代入します。

## 3. 今回覚えるポイント

- 列の追加: `df["new"] = 式`
- 列どうしの `+ - * /` は行ごとにまとめて計算される。**for 文や apply() は不要**
- どちらかが欠損の行は、結果も欠損になる

## 4. よくある間違い

- `df.amount = ...` → 新しい列は作られない。必ず `df["amount"]` で代入する
- `df.apply(lambda row: row["price"] * row["quantity"], axis=1)` → 動くが遅く読みにくい。このカリキュラムでは使わない
