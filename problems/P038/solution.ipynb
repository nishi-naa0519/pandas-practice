# P038: testにだけ現れるカテゴリを見つける

**解答と解説**

```python
import pandas as pd
from src.grader import check, hint

train = pd.read_csv("data/churn_train.csv")
test = pd.read_csv("data/churn_test.csv")
```

## 1. 模範解答

```python
def solve(train: pd.DataFrame, test: pd.DataFrame):
    return set(test["region"]) - set(train["region"])
```

```python
solve(train, test)
```

```python
check("P038", solve, record=False)  # 解答例の確認用(履歴には記録しない)
```

## 2. コードの説明

`set(test["region"])` で test に出てくる region の集合を作り、train の集合を引きます。
残ったものが「test にだけある値」です。

## 3. 今回覚えるポイント

- カテゴリ集合の比較: `set(test["col"]) - set(train["col"])`
- 逆向き(train にだけある値)も `set(train["col"]) - set(test["col"])` で確認できる
- test にだけあるカテゴリは、学習時に見たことがない値。エンコードでエラーになったり、予測が不安定になったりする
- 列名の比較も同じ: `set(train.columns) - set(test.columns)`

## 4. よくある間違い

- 引き算の順番を逆にする → train にだけある値になる
- 欠損(NaN)を含む列では、NaN が差集合に残ることがある。気になったら `value_counts()` で中身を見る
