```python
import pandas as pd
from src.grader import check, hint

df = pd.read_csv("data/store_sales.csv")
```

## 1. 模範解答

```python
def solve(df: pd.DataFrame):
    return df.sort_values("revenue", ascending=False).head(5)
```

```python
solve(df)
```

```python
check("P021", solve, record=False)  # 解答例の確認用(履歴には記録しない)
```

## 2. コードの説明

1. `sort_values("revenue", ascending=False)` で revenue の大きい順に並べる
2. `.head(5)` で上から5行を取る

欠損(NaN)は並び替えると最後に回るので、上位には出てきません。

## 3. 今回覚えるポイント

- 上位N件: `df.sort_values("col", ascending=False).head(n)`
- 下位N件は `ascending=True`(デフォルト)のまま `.head(n)`
- 異常に大きい値・小さい値を探すときにも、まず並べて見る

## 4. よくある間違い

- `ascending=False` を忘れて小さい順になる
- `df.sort_values("revenue", ascending=False)` だけで代入しない → `df` 自体は並び替わらない(結果を使う)
- `df["revenue"].sort_values(...)` → revenue 列だけになり、他の列が見えない
