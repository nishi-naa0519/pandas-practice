```python
import pandas as pd
from src.grader import check, hint

df = pd.read_csv("data/customers.csv")
```

## 1. 模範解答

```python
def solve(df: pd.DataFrame):
    return df.duplicated().sum()
```

```python
solve(df)
```

```python
check("P017", solve, record=False)  # 解答例の確認用(履歴には記録しない)
```

## 2. コードの説明

`df.duplicated()` は、すべての列の値が**それより前の行と完全に一致する行**を True にします。
1回目の出現は False なので、`.sum()` で「2回目以降の行の数」になります。

## 3. 今回覚えるポイント

- 重複行の数: `df.duplicated().sum()`
- ID列だけの重複: `df["customer_id"].duplicated().sum()`(他の列が違っていても数える)
- 重複行の中身を見る: `df[df.duplicated()]`

## 4. よくある間違い

- 一部の列だけ同じ行は、`df.duplicated()` では重複にならない
- `len(df) - df.nunique()` などは別物
