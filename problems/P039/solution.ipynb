**解答と解説**

```python
import pandas as pd
from src.grader import check, hint

df = pd.read_csv("data/churn_train.csv")
```

## 1. 模範解答

```python
def solve(df: pd.DataFrame):
    n_unique = df.nunique()
    return list(n_unique[n_unique >= 50].index)
```

```python
solve(df)
```

```python
check("P039", solve, record=False)  # 解答例の確認用(履歴には記録しない)
```

## 2. コードの説明

1. `df.nunique()` で「列名 → ユニーク数」の Series を作る
2. `n_unique[n_unique >= 50]` で、DataFrame と同じように条件で絞る
3. 残った Series の index(= 列名)をリストにする

## 3. 今回覚えるポイント

- 列ごとのユニーク数: `df.nunique()`
- Series も `s[s >= 50]` で絞れる。`.index` で列名が取れる
- ユニーク数が行数に近い列は ID の可能性が高い(このデータの `id`, `postal_code`)
- 数値の列(`age` など)はユニーク数が多くて当然。カテゴリ列かどうかと合わせて判断する

## 4. よくある間違い

- `df[df.nunique() >= 50]` → 行の抽出になってしまう
- `.index` ではなく `.values` → ユニーク数の値が返る
