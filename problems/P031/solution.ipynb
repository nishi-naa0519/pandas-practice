**解答と解説**

```python
import pandas as pd
from src.grader import check, hint

df = pd.read_csv("data/orders.csv")
```

## 1. 模範解答

```python
def solve(df: pd.DataFrame):
    df["order_date"] = pd.to_datetime(df["order_date"])
    df["year"] = df["order_date"].dt.year
    df["month"] = df["order_date"].dt.month
    return df
```

```python
solve(df)
```

```python
check("P031", solve, record=False)  # 解答例の確認用(履歴には記録しない)
```

## 2. コードの説明

1. `read_csv` は日付を文字列のまま読み込むので、`pd.to_datetime()` で日時型にします
2. 日時型の列は `.dt.year`, `.dt.month`, `.dt.day` などで要素を取り出せます

## 3. 今回覚えるポイント

- 変換: `df["date"] = pd.to_datetime(df["date"])`
- 取り出し: `.dt.year` / `.dt.month` / `.dt.day`(`.dt.dayofweek` で曜日 0=月曜)
- 変換後は `df["date"].min()` / `.max()` で期間も確認できる

## 4. よくある間違い

- 文字列のまま `.dt.year` → `AttributeError: Can only use .dt accessor with datetimelike values`
- `df["order_date"].dt.year()` → 括弧は付けない(属性)
- `df["order_date"].to_datetime()` → `pd.to_datetime(...)` は pd の関数
