```python
import pandas as pd
from src.grader import check, hint

df = pd.read_csv("data/customers.csv")
```

## 1. 模範解答

```python
def solve(df: pd.DataFrame):
    return df.isna().sum()
```

```python
solve(df)
```

```python
check("P014", solve, record=False)  # 解答例の確認用(履歴には記録しない)
```

## 2. コードの説明

1. `df.isna()` で、欠損のセルが True、それ以外が False の DataFrame を作ります
2. `.sum()` は列ごとに合計します。True は 1、False は 0 として数えられるので、欠損数になります

## 3. 今回覚えるポイント

- 欠損数: `df.isna().sum()`
- 1列だけ: `df["age"].isna().sum()`
- True/False の Series に `.sum()` → True の件数、という使い方はいろいろな場面で出てくる

## 4. よくある間違い

- `df.isna` と括弧を忘れる
- `df.count()` は**欠損でない**件数。欠損数とは逆
