## 2. コードの説明

`df.dtypes` は「列名 → 型」の Series です。

このデータでは `age` が `float64` になっています。年齢は整数のはずですが、
**欠損(NaN)を含む整数列は float になる**ためです。

## 3. 今回覚えるポイント

- 型の確認は `df.dtypes`
- よく見る型: `int64`(整数)/ `float64`(小数・欠損を含む整数)/ `str`(文字列)/ `bool` / `datetime64`
- `read_csv` は日付を自動では日時型にしない → 日付列は `str` のまま
- `"0600001"` のような郵便番号やIDは数値として読まれて**先頭の0が消える**ことがある
- 「数値のはずなのに `str`」は、どこかに数値でない値("unknown" など)が混ざっているサイン

## 4. よくある間違い

- `df.dtype` → DataFrame にはない(Series なら `df["age"].dtype`)
- `df.dtypes()` → 括弧は不要
