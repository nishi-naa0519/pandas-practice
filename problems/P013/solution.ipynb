**解答と解説**

```python
import pandas as pd
from src.grader import check, hint

df = pd.read_csv("data/web_logs.csv")
```

## 1. 模範解答

```python
def solve(df: pd.DataFrame):
    return df.iloc[:5, :3]
```

```python
solve(df)
```

```python
check("P013", solve, record=False)  # 解答例の確認用(履歴には記録しない)
```

## 2. コードの説明

`df.iloc[行, 列]` は位置(0始まりの番号)で指定します。`:5` は 0〜4 番目で、**終わりの位置は含みません**。

## 3. 今回覚えるポイント

- `loc` はラベル(条件・列名)、`iloc` は位置(番号)
- `df.iloc[0]` で1行目、`df.iloc[-1]` で最後の行を Series で見られる

## 4. よくある間違い

- `df.iloc[:5, ["session_id", ...]]` → iloc に列名は使えない
- `df.loc[:5, :3]` → loc はラベル指定なので意味が変わる
