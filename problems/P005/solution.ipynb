## 2. コードの説明

`df.describe()` は数値列ごとに count / mean / std / min / 25% / 50%(中央値) / 75% / max を計算します。
文字列の列は自動で除外されます。

## 3. 今回覚えるポイント

`describe()` は「眺めて気づく」ための道具です。見るポイント:
- **count** が行数より少ない → 欠損がある(このデータでは `revenue`)
- **mean と 50% の差**が大きい → 分布が歪んでいる・極端な値がある
- **min / max** → 負の値、ありえない大きな値がないか

## 4. よくある間違い

- 文字列の列が結果に出てこないのは正常です
- 欠損は除外して計算されるので、count が少ない列の平均は「欠損以外の平均」です
