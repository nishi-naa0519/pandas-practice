**解答と解説**

```python
import pandas as pd
from src.grader import check, hint

df = pd.read_csv("data/web_logs.csv")
```

## 1. 模範解答

```python
def solve(df: pd.DataFrame):
    human = df[~df["device"].isin(["bot", "crawler"])]
    return human["user_id"].value_counts().head(3)
```

```python
solve(df)
```

```python
check("P041", solve, record=False)  # 解答例の確認用(履歴には記録しない)
```

## 2. コードの説明

1. `df["device"].isin(["bot", "crawler"])` は bot / crawler なら True
2. `~` で反転して「bot でも crawler でもない」行に絞る
3. `user_id` の `value_counts()`(多い順)から `head(3)`

## 3. 今回覚えるポイント

- 除外: `df[~df["col"].isin([...])]`
- ボットや社内テストアカウントなど、**分析対象でない行を先に除く**のは EDA の定番
- `value_counts()` は多い順に並ぶので、`head(n)` で上位N件になる

## 4. よくある間違い

- `~` を付け忘れる → bot と crawler だけが残る
- `df["device"] != ["bot", "crawler"]` → エラー。複数の値には isin
